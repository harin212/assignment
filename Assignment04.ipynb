### 225. Implement Stack using Queues
##### 두 개의 큐를 사용하여 스택 기능을 구현

```python
from collections import deque

class MyStack(object):

    def __init__(self):
        self.in_queue = deque() # 입력을 위한 큐
        self.out_queue = deque() # 출력용 큐

    def push(self, x:int) -> None:
        self.in_queue.append(x) # 'in_queue'에 새 항목 추가

    def pop(self) -> int:
        # 'in_queue'에 있는 모든 요소를 'out_queue'로 이동시키고, 마지막에 남은 항목 pop
        while len(self.in_queue) > 1:
            self.out_queue.append(self.in_queue.popleft())
        # 마지막 요소를 pop하고, 'in_queue'를 비움
        top = self.in_queue.popleft()
        self.in_queue, self.out_queue = self.out_queue, self.in_queue
        return top

    def top(self) -> int:
        # 'in_queue'에 있는 요소를 pop하지 않고 top 요소를 가져옴
        while len(self.in_queue) > 1:
            self.out_queue.append(self.in_queue.popleft())
        top = self.in_queue.popleft()
        self.out_queue.append(top) # pop한 항목을 다시 'out_queue'에 넣음
        self.in_queue, self.out_queue = self.out_queue, self.in_queue # 큐들 교환
        return top

    def empty(self) -> bool:
        # 'in_queue'와 'out_queue'가 모두 비었으면 스택이 비어있다고 판단
        return not self.in_queue
```

```python
# test
stack = MyStack()
stack.push(1)
stack.push(2)
print(stack.top()) # 예상:2
print(stack.pop()) # 예상:2
print(stack.empty()) # 예상:False
```

### 232. Implement Queue using Stacks
##### 두 개의 스택을 이용해서 큐 기능을 구현

```python
class MyQueue:

    def __init__(self):
        self.stack1 = [] # 큐에 데이터를 넣는 데 사용
        self.stack2 = [] # 큐에서 데이터를 꺼내는 데 사용

    def push(self, x: int) -> None:
        self.stack1.append(x) # 'stack1'에 새 항목 추가

    def pop(self) -> int:
        # 'stack2'가 비어있다면, 'stack1'에 있는 모든 요소를 'stack2'로 이동시키고, 'stack2' pop()
        if not self.stack2:
            while self.stack1:
                self.stack2.append(self.stack1.pop())
        # 'stack2'를 비움
        return self.stack2.pop()

    def peek(self) -> int:
        # 'stack2'가 비어있다면, 'stack1'에 있는 모든 요소를 'stack2'로 이동시키고, 'stack2'의 top 확인
        if not self.stack2:
            while self.stack1:
                self.stack2.append(self.stack1.pop())
        return self.stack2[-1]

    def empty(self) -> bool:
        # 'stack1'과 'stack2'가 모두 비었으면 스택이 비어있다고 판단
        return not self.stack1 and not self.stack2
```

```python
# test
queue = MyQueue()
queue.push(1)
queue.push(2)
print(queue.peek()) # 예상:1
print(queue.pop()) # 예상:1
print(queue.empty()) # 예상:False
```

##### 2.

```python
from collections import deque

def is_in_language(s):
    q1 = deque()
    q2 = deque()

    if '$' not in s:
        return False

    left, right = s.split('$', 1)

    # 큐에 각각 문자 넣기
    for char in left:
        q1.append(char)
    for char in right:
        q2.append(char)

    # 큐에서 하나씩 꺼내면서 비교
    while q1 and q2:
        if q1.popleft() != q2.popleft():
            return False

    # 둘 다 비었으면 같다는 뜻
    return not q1 and not q2
```

```python
# test
print(is_in_language("abc$abc")) # 예상:True
print(is_in_language("abc$cba")) # 예상:False
print(is_in_language("123$123")) # 예상:True
```

```python
class Node:
    def __init__(self, data, next=None):
        self.data = data # 이 노드가 저장하는 실제 데이터
        self.next = next # 다음 노드를 가리키는 참조 (없으면 None)

class LinkedQueue:
    def __init__(self):
        self.front = None # 큐의 맨 앞 노드
        self.rear = None # 큐의 맨 뒤 노드

    def is_empty(self):
        return self.front is None # 큐가 비었는지 확인. front가 없으면 True 반환

    def enqueue(self, x):
        new_node = Node(x) # 큐의 뒤(rear)에 새 데이터 추가
        if self.is_empty():
            self.front = self.rear = new_node # 큐가 비어있다면 front와 rear가 모두 이 노드를 가리킴
        else:
            self.rear.next = new_node # 아니면 현재 rear 뒤에 새 노드 연결, rear를 새 노드로 업데이트
            self.rear = new_node

    def dequeue(self):
        if self.is_empty():
            raise IndexError("Queue is empty") # 큐가 비어있으면 에러 발생
        data = self.front.data # 큐의 앞(front)에서 데이터를 꺼냄
        self.front = self.front.next # 데이터를 꺼낸 뒤 front를 다음 노드로 이동
        if self.front is None:
            self.rear = None # 꺼낸 후 큐가 비면 rear도 None으로 바꿈
        return data

    def __iter__(self):
        current = self.front
        while current:
            yield current.data # 데이터를 하나씩 전달
            current = current.next

    def copy(self):
        new_queue = LinkedQueue()
        for x in self:
            new_queue.enqueue(x)
        return new_queue
```

```python
# 사용 예시
a = LinkedQueue()
a.enqueue(1)
a.enqueue(2)
a.enqueue(3)

b = a.copy()

# a와 b는 서로 다른 객체지만 내용은 동일함
print(list(a))  # [1, 2, 3]
print(list(b))  # [1, 2, 3]

# b를 바꿔도 a는 영향 없음
b.enqueue(4)
print(list(a))  # [1, 2, 3]
print(list(b))  # [1, 2, 3, 4]
```

```python
class MyQueue:

    def __init__(self):
        self.stack1 = [] # 큐에 데이터를 넣는 데 사용
        self.stack2 = [] # 큐에서 데이터를 꺼내는 데 사용

    def push(self, x: int) -> None:
        self.stack1.append(x) # 'stack1'에 새 항목 추가

    def pop(self) -> int:
        # 'stack2'가 비어있다면, 'stack1'에 있는 모든 요소를 'stack2'로 이동시키고, 'stack2' pop()
        if not self.stack2:
            while self.stack1:
                self.stack2.append(self.stack1.pop())
        # 'stack2'를 비움
        return self.stack2.pop()
```

##### 6.

[코드 7-13] 의 LinkedQueue는 내부적으로 CircularLinkedList를 사용한다
- enqueue(x) -> self.__queue.append(x)
- dequeue() -> self.__queue.pop(0)

*CircularLinkedList는 양방향 연결 리스트 기반*

수행 시간:
- append(x) -> O(1) (tail 포인터 유지 시)
- pop(0) → O(1) (head 포인터 유지 시)

따라서
- enqueue() → O(1)
- dequeue() → O(1)

##### 7.

[코드 5-15] 의 LinkedListBasic은 단일 연결 리스트
- append(x) → tail이 없기 때문에 마지막 노드까지 순회: O(n)
- pop(0) → 첫 노드 제거는 head에서 바로 가능: O(1)

따라서
- enqueue() → O(n)
- dequeue() → O(1)

```python
class Deque:
    def __init__(self):
        self.__deque = []

    def enqueueFront(self, x):
        self.__deque.insert(0, x)  # 리스트 앞에 삽입 (O(n))

    def enqueueRear(self, x):
        self.__deque.append(x)     # 리스트 뒤에 삽입 (O(1))

    def dequeueFront(self):
        if self.isEmpty():
            return None
        return self.__deque.pop(0)  # 리스트 앞에서 삭제 (O(n))

    def dequeueRear(self):
        if self.isEmpty():
            return None
        return self.__deque.pop()   # 리스트 뒤에서 삭제 (O(1))

    def front(self):
        if self.isEmpty():
            return None
        return self.__deque[0]

    def rear(self):
        if self.isEmpty():
            return None
        return self.__deque[-1]

    def isEmpty(self):
        return len(self.__deque) == 0

    def dequeueAll(self):
        self.__deque.clear()

    def printDeque(self):
        print("Deque from front:", end=' ')
        for item in self.__deque:
            print(item, end=' ')
        print()
```

```python
# 사용 예시
dq = Deque()
dq.enqueueRear(10)
dq.enqueueFront(20)
dq.enqueueRear(30)
dq.printDeque()         # 출력: Deque from front: 20 10 30

print(dq.dequeueFront()) # 20
print(dq.dequeueRear())  # 30
dq.printDeque()          # 출력: Deque from front: 10
```
